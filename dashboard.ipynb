# Enterprise Sales Intelligence: Exploratory Data Analysis (EDA)

This notebook documents the Day 1 to Day 5 development of the Interactive Sales Dashboard. 
The objective is to prototype our statistical and interactive visualizations before porting them to the Streamlit web framework.

**Goals:**
* Generate and validate the mock enterprise dataset.
* Prototype Seaborn static plots (Boxplot, Violin, Heatmap).
* Prototype Plotly interactive components (Line, Treemap).
* Export static assets for documentation.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px

# Enforcing the Modern Innovator Theme in Jupyter
BG_COLOR = "#121212"
TEXT_COLOR = "#EAEAEA"
ACCENT_BLUE = "#3366FF"
HIGHLIGHT_GREEN = "#1DB954"
ACCENT_PURPLE = "#8A2BE2"

plt.style.use('dark_background')
plt.rcParams.update({
    "figure.facecolor": BG_COLOR,
    "axes.facecolor": BG_COLOR,
    "text.color": TEXT_COLOR,
    "axes.labelcolor": TEXT_COLOR,
    "xtick.color": TEXT_COLOR,
    "ytick.color": TEXT_COLOR,
    "axes.edgecolor": "#333333",
})
sns.set_palette([ACCENT_BLUE, HIGHLIGHT_GREEN, ACCENT_PURPLE])

# Inline rendering for Jupyter
%matplotlib inline
```

```python
# Day 1: Data Generation
np.random.seed(42)
categories = ['Electronics', 'Software', 'Hardware', 'Services']
data = {
    'Date': pd.date_range(start='2025-01-01', periods=365),
    'Category': np.random.choice(categories, 365),
    'Price': np.random.normal(150, 40, 365),
    'Units_Sold': np.random.randint(1, 100, 365),
    'Customer_Age': np.random.normal(35, 10, 365),
    'Discount': np.random.uniform(0, 0.3, 365)
}
df = pd.DataFrame(data)
df['Revenue'] = df['Price'] * df['Units_Sold'] * (1 - df['Discount'])

# Validate data quality (Testing for nulls)
assert df.isnull().sum().sum() == 0, "Data contains null values"
df.head()
```

```python
# Day 2: Boxplot & Violin Plot Prototyping
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

sns.boxplot(x='Category', y='Price', data=df, ax=ax1, width=0.5)
ax1.set_title("Distribution Check: Price by Category")

sns.violinplot(x='Category', y='Customer_Age', data=df, ax=ax2, inner="quartile")
ax2.set_title("Demographics Check: Age by Category")

plt.show()
```

```python
# Day 3: Heatmap Prototyping
plt.figure(figsize=(8, 6))
corr = df[['Price', 'Units_Sold', 'Customer_Age', 'Discount', 'Revenue']].corr()

# Custom dark theme mapping for the heatmap
sns.heatmap(corr, annot=True, cmap=sns.dark_palette(ACCENT_BLUE, as_cmap=True), 
            cbar_kws={'label': 'Correlation Coefficient'})
plt.title("Feature Correlation Matrix")
plt.show()
```

```python
# Day 5: Interactive Visualizations Prototyping
daily_rev = df.groupby('Date')['Revenue'].sum().reset_index()
fig_line = px.line(daily_rev, x='Date', y='Revenue', title="Daily Revenue Trend")

fig_line.update_traces(line_color=HIGHLIGHT_GREEN, line_width=3)
fig_line.update_layout(plot_bgcolor=BG_COLOR, paper_bgcolor=BG_COLOR, font_color=TEXT_COLOR)

# This will render interactively right inside the Jupyter Notebook cell
fig_line.show()
```
